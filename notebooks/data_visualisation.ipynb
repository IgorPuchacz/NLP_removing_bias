```
# Import
import pandas as pd
from pathlib import Path

# Fetch data
project_root = Path.cwd().parent
data_file    = project_root / "data" / "dataset_6.csv"
df = pd.read_csv(data_file)
print("🔢 Number of examples:", len(df))
df.head()
```

```
# Counts
label_counts = pd.Series({
    "stereotype": df["stereotype"].notna().sum(),
    "anti-stereotype": df["anti-stereotype"].notna().sum(),
    "meaningless": df["meaningless"].notna().sum()
})
bias_counts = df['label'].value_counts()

print("🟦 Class count:\n", label_counts)
print("\n🟨 Bias type count:\n", bias_counts)

# TODO rozklad przez genders
```
