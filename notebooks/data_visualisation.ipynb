```
# Import
import pandas as pd
from pathlib import Path

# Fetch data
project_root = Path.cwd().parent
data_file    = project_root / "data" / "dataset_6.csv"
df = pd.read_csv(data_file)
print("🔢 Number of examples:", len(df))
df.head()
```

```
# Counts
male_count   = (df['label'] == 0).sum()
female_count = (df['label'] == 1).sum()

print("🔢 Number of examples:", len(df))
print("👨🏽 Male examples:   ", male_count)
print("👩🏽 Female examples: ", female_count)
```
